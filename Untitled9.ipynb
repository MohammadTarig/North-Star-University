````python
# Content Scaffolding Module - AI-Powered Course Structure Generator
# Using Google Gemini API - CORRECTED VERSION

import json
import time
import google.generativeai as genai

# Install required package
!pip install -q google-generativeai

# Configure Google Gemini - USE YOUR GOOGLE API KEY
genai.configure(api_key="AIzaSyCViKKJOLPAAvh-whzRplf6zyw-yX4E-H0")

def validate_course_structure(output_json):
    """Validate if the generated JSON matches the required schema"""
    required_fields = ["topic", "objectives", "modules", "project_ideas"]

    for field in required_fields:
        if field not in output_json:
            return False, f"Missing required field: '{field}'"

    return True, "Valid JSON structure"

def generate_course_scaffolding(topic):
    """Generate course structure using Google Gemini"""

    prompt_template = f"""
    IMPORTANT: Output ONLY valid JSON format, no additional text.

    Create a course structure for '{topic}' with this exact JSON format:
    {{
        "topic": "topic name here",
        "objectives": ["objective 1", "objective 2", "objective 3"],
        "modules": [
            {{
                "title": "module title 1",
                "concepts": ["concept 1", "concept 2", "concept 3"]
            }},
            {{
                "title": "module title 2",
                "concepts": ["concept 1", "concept 2"]
            }}
        ],
        "project_ideas": ["project idea 1", "project idea 2"]
    }}

    Requirements:
    - 3-5 learning objectives
    - 3-5 modules with titles and concepts
    - 2-3 project ideas
    - Logical progression from basic to advanced

    Generate for: {topic}
    """

    try:
        # Use the correct model name and configuration
        model = genai.GenerativeModel('gemini-1.5-flash')

        response = model.generate_content(
            prompt_template,
            generation_config=genai.types.GenerationConfig(
                temperature=0.7,
                max_output_tokens=1000,
            )
        )

        content = response.text.strip()

        # Clean the response and extract JSON
        content = content.replace('```json', '').replace('```', '').strip()

        # Extract JSON from response
        start_idx = content.find('{')
        end_idx = content.rfind('}') + 1

        if start_idx == -1 or end_idx == 0:
            print("No JSON found in response")
            return None

        json_str = content[start_idx:end_idx]
        return json.loads(json_str)

    except Exception as e:
        print(f"API error: {e}")
        return None

def main():
    """Main function to generate course structures"""
    print("=== AI-Powered Course Scaffolding Generator ===")
    print("Using Google Gemini API - Corrected Version")
    print("=" * 50)

    # Test with simple topics first
    test_topics = ["Loops in Python", "Photosynthesis"]
    max_retry_attempts = 2

    for topic in test_topics:
        print(f"\n📚 Generating structure for: {topic}")
        print("-" * 40)

        success = False
        for attempt in range(max_retry_attempts):
            print(f"🔄 Attempt {attempt + 1}/{max_retry_attempts}")

            course_structure = generate_course_scaffolding(topic)

            if course_structure:
                is_valid, validation_message = validate_course_structure(course_structure)

                if is_valid:
                    filename = f"{topic.lower().replace(' ', '_')}_course.json"
                    with open(filename, 'w', encoding='utf-8') as f:
                        json.dump(course_structure, f, ensure_ascii=False, indent=2)

                    print(f"✅ SUCCESS: Structure generated and validated!")
                    print(f"💾 Saved to: {filename}")

                    # Display preview
                    print(f"📊 Preview:")
                    print(f"   - Topic: {course_structure['topic']}")
                    print(f"   - Objectives: {len(course_structure['objectives'])}")
                    print(f"   - Modules: {len(course_structure['modules'])}")
                    print(f"   - Projects: {len(course_structure['project_ideas'])}")

                    success = True
                    break
                else:
                    print(f"❌ Validation failed: {validation_message}")
                    # Try to fix the JSON or use a simpler approach
            else:
                print("❌ Generation failed, retrying...")

            time.sleep(1)  # Short delay between attempts

        if not success:
            print(f"⚠️  Using fallback method for: {topic}")
            # Create a simple fallback structure
            fallback_structure = {
                "topic": topic,
                "objectives": [f"Understand basic concepts of {topic}", f"Apply {topic} in practical scenarios", f"Master advanced techniques in {topic}"],
                "modules": [
                    {
                        "title": f"Introduction to {topic}",
                        "concepts": ["Basic concepts", "Fundamental principles", "Key terminology"]
                    },
                    {
                        "title": f"Intermediate {topic}",
                        "concepts": ["Advanced techniques", "Practical applications", "Common patterns"]
                    }
                ],
                "project_ideas": [f"Basic {topic} project", f"Advanced {topic} application"]
            }

            filename = f"{topic.lower().replace(' ', '_')}_course.json"
            with open(filename, 'w', encoding='utf-8') as f:
                json.dump(fallback_structure, f, ensure_ascii=False, indent=2)

            print(f"📁 Fallback structure saved to: {filename}")

    print("\n" + "=" * 50)
    print("🎉 TASK COMPLETED SUCCESSFULLY!")
    print("📂 Generated files:")
    for topic in test_topics:
        filename = f"{topic.lower().replace(' ', '_')}_course.json"
        print(f"   - {filename}")
    print("=" * 50)

# Run the program
main()
````
